# Image Crop Notebook

A Jupyter notebook that crops photos to display both faces while maintaining image quality.

```python
## 1. Import Required Libraries

Import libraries for image processing (OpenCV, PIL), face detection (face_recognition), and visualization (matplotlib).
```

```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
from PIL import Image

try:
    import face_recognition
except ImportError:
    raise ImportError(
        "face_recognition is required for face detection. Install it via `pip install face_recognition`"
    )

IMAGE_PATH = Path("images/couple.jpg")
assert IMAGE_PATH.exists(), f"Image not found: {IMAGE_PATH}"
```

## 2. Load and Display the Original Image

Load the image file and display it using matplotlib to inspect the original photo and face positions.

```python
original_image = Image.open(IMAGE_PATH).convert("RGB")
plt.figure(figsize=(8, 10))
plt.imshow(original_image)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

## 3. Detect Faces in the Image

Use a face detection library to identify all faces in the image and extract their bounding box coordinates.

```python
image_array = face_recognition.load_image_file(IMAGE_PATH)
face_locations = face_recognition.face_locations(image_array)
print('Detected face locations:', face_locations)

if not face_locations:
    raise RuntimeError('No faces detected in the image.')

plt.figure(figsize=(8, 10))
plt.imshow(image_array)
for top, right, bottom, left in face_locations:
    cv2.rectangle(image_array, (left, top), (right, bottom), (255, 0, 0), 3)
plt.axis('off')
plt.title('Detected Faces')
plt.show()
```

## 4. Calculate Optimal Crop Coordinates

Compute the minimal bounding rectangle that encompasses all detected faces with appropriate padding to maintain composition and quality.

```python
height, width, _ = image_array.shape
face_top = min(top for top, _, _, _ in face_locations)
face_right = max(right for _, right, _, _ in face_locations)
face_bottom = max(bottom for _, _, bottom, _ in face_locations)
face_left = min(left for _, _, _, left in face_locations)

padding = int(max(width, height) * 0.08)

crop_left = max(0, face_left - padding)
crop_top = max(0, face_top - padding)
crop_right = min(width, face_right + padding)
crop_bottom = min(height, face_bottom + padding)

crop_width = crop_right - crop_left
crop_height = crop_bottom - crop_top

if crop_width > crop_height:
    delta = crop_width - crop_height
    crop_top = max(0, crop_top - delta // 2)
    crop_bottom = min(height, crop_bottom + delta - delta // 2)
else:
    delta = crop_height - crop_width
    crop_left = max(0, crop_left - delta // 2)
    crop_right = min(width, crop_right + delta - delta // 2)

crop_left, crop_top, crop_right, crop_bottom, crop_right - crop_left, crop_bottom - crop_top
```

## 5. Crop the Image

Extract the cropped region from the original image using the calculated coordinates while preserving the original resolution and color quality.

```python
cropped_image = original_image.crop((crop_left, crop_top, crop_right, crop_bottom))
out_path = Path('images/couple_cropped.jpg')
cropped_image.save(out_path, quality=95)
print('Saved cropped image to', out_path)
```

## 6. Compare Original and Cropped Images

Display both the original and cropped images side-by-side to verify that both faces are visible and the crop maintains image quality.

```python
plt.figure(figsize=(14, 8))
plt.subplot(1, 2, 1)
plt.imshow(original_image)
plt.axis('off')
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cropped_image)
plt.axis('off')
plt.title('Cropped Image')

plt.show()
```
